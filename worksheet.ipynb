Properties of .gz file

```python
import SimpleITK as sitk

img = sitk.ReadImage(r"C:\Users\rajaj\Downloads\Radiology\TCIA_sample_cases\3.Visualization_code\outputs\highest_bvalue_series.nii.gz")
print("Size (voxels):", img.GetSize())          # (columns, rows, slices)
print("Spacing (mm):", img.GetSpacing())
print("Origin:", img.GetOrigin())
print("Direction:", img.GetDirection())

import numpy as np
arr = sitk.GetArrayFromImage(img)
print("Array shape:", arr.shape)                 # (slices, rows, columns)
print("Intensity range:", arr.min(), "-", arr.max())
```
